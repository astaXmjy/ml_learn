Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
x=np.linspace(0,10,100)
y=np.sin(x)
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin(x)')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r')
plt.subplot(1,2,2)
plt.plot(x,y,'b')
```

```python
#OO
fig=plt.figure()
axes=fig.add_axes([0.5,0.5,0.9,0.9])
axes.plot(x,y)
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('sin(x)')
```

```python
fig=plt.figure()

axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0,0,0.4,0.3])
axes1.plot(x,y)
axes2.plot(y,x)
```

Plot Appearance

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
x=[1,2,3,4,5]
y=[x**2 for x in x]
ax.plot(x,y,color='purple',linewidth=3,alpha=1,linestyle='-.')
ax.set_xlim([0,10])
ax.set_ylim([0,30])
```
