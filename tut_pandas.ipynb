```python
data={
    'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
    'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
    'Sales':[200,120,340,124,243,350]
}
```

```python
data
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
byComp=df.groupby('Company')
```

```python
byComp['Sales'].mean()
```

```python
byComp.sum()
```

```python
byComp['Sales'].std()
```

```python
byComp.sum().loc['FB']
```

```python
df.groupby('Company').count()
```

```python
df.groupby('Company').max()
```
